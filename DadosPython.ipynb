```python
df.rename(columns={
    'work_year': 'ano',
    'experience_level': 'senioridade',
    'employment_type': 'contrato',
    'job_title': 'cargo',
    'salary': 'salario',
    'salary_currency': 'moeda',
    'salary_in_usd': 'usd',
    'employee_residence': 'residencia',
    'remote_ratio': 'remoto',
    'company_location': 'empresa',
    'company_size': 'tamanho_da_empresa'
}, inplace=True)

df.columns
```

```python
df['senioridade'].value_counts()
```

```python
df['contrato'].value_counts()
```

```python
df['remoto'].value_counts()
```

```python
df['tamanho_da_empresa'].value_counts()
```

```python
senioridade = {
    'SE': 'senior',
    'MI': 'pleno',
    'EN': 'junior',
    'EX': 'executivo'
}
df['senioridade'] = df['senioridade'].replace(senioridade)
df['senioridade'].value_counts()
```

```python
contrato = {
    'FT': 'integral',
    'PT': 'parcial',
    'CT': 'contrato',
    'FL': 'freelancer'
}
df['contrato'] = df['contrato'].replace(contrato)
df['contrato'].value_counts()
```

```python
tamanho_da_empresa = {
    'L': 'grande',
    'S': 'pequena',
    'M':	'media'

}
df['tamanho_da_empresa'] = df['tamanho_da_empresa'].replace(tamanho_da_empresa)
df['tamanho_da_empresa'].value_counts()
```

```python
remoto = {
    0: 'presencial',
    100: 'remoto',
    50: 'hibrido'
}

df['remoto'] = df['remoto'].replace(remoto)
df['remoto'].value_counts()
```

```python
#extenção para colocar informações nulas
import numpy as np

#Criando um data frame para exemplo de preenchimento de informação isnull com média
df_salarios = pd.DataFrame({
    'nome':['Ana','Bruno','Carlos','Daniele', 'Val'],
    'salario':[4000, np.nan, 5000, np.nan, 100000]
})

#Código para criar nova coluna e preencher os nulos com média de salarios.
df_salarios['salarios_media'] = df_salarios['salario'].fillna(df_salarios['salario'].mean().round(2))

#Calcula a mediana e preenche os nulos com a mediana


df_salarios['salarios_mediana'] = df_salarios['salario'].fillna(df_salarios['salario'].median())

df_salarios
```

```python
df_temperaturas = pd.DataFrame({
    'Dia': ['Segunda', 'Terça', 'Quarta', 'Quinta', 'Sexta'],
    'Temperatura': [30, np.nan, np.nan, 28, 27]
})

#Configurando código com comando ffill , preenchendo espaços nulos com informações anteriores.
df_temperaturas['Preenchi_ffill'] = df_temperaturas['Temperatura'].ffill()
df_temperaturas
```

```python
df_temperaturas = pd.DataFrame({
    'Dia': ['Segunda', 'Terça', 'Quarta', 'Quinta', 'Sexta'],
    'Temperatura': [30, np.nan, np.nan, 28, 27]
})

#Configurando código com comando bfill , preenchendo espaços nulos com informações posteriores.
df_temperaturas['Preenchi_bfill'] = df_temperaturas['Temperatura'].bfill()
df_temperaturas
```

```python
df_cidades = pd.DataFrame({
    'Nome':['Ana','Bruno','Carlos','Daniele', 'Val'],
    'Cidades': ['São Paulo', np.nan, 'Curitiba', np.nan, 'Belém']
})

#Usando comando fillna para preenchimento de espaços nulos com uma informação definida.
df_cidades['cidade_preenchida'] = df_cidades['Cidades'].fillna('Não informado')
df_cidades
```

```python
import matplotlib.pyplot as plt
```

```python
#utilizando matplotlib, junto com seaborn e pandas para melhorar visualização dos dados
plt.figure(figsize=(8,5))
sns.barplot(data=df_limpo, x='senioridade', y='usd')
plt.title('Sálario médio por senioridade')
plt.xlabel('Senioridade')
plt.ylabel('Sálario médio anual (USD)')
plt.show()
```

```python
#código para organizar de maior para menores valores
df_limpo.groupby('senioridade')['usd'].mean().sort_values(ascending=False)
```

```python
#nomeando variavel para utilização no código do gráfico
ordem = df_limpo.groupby('senioridade')['usd'].mean().sort_values(ascending=False).index
ordem
```

```python
#Adicionando parametro order para plotar gráfico em ordem de maior para menor
plt.figure(figsize=(8,5))
sns.barplot(data=df_limpo, x='senioridade', y='usd', order=ordem)
plt.title('Sálario médio por senioridade')
plt.xlabel('Senioridade')
plt.ylabel('Sálario médio anual (USD)')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(df_limpo['usd'], bins = 50, kde=True)
plt.title('Distribuição dos sálarios anuais')
plt.xlabel('Sálario em USD')
plt.ylabel('Frequência')
plt.show()
```

```python
#gráfico bloxplot mostrando média, taxa de 25%, 75% e exceções.
plt.figure(figsize=(8,5))
sns.boxplot(x=df_limpo['usd'])
plt.title('Boxplot Sálario')
plt.xlabel('Sálario em USD')
plt.show()
```

```python
#Boxplot por categoria
ordem_senioridade = ['junior', 'pleno', 'senior', 'executivo']
plt.figure(figsize=(8,5))
sns.boxplot(x='senioridade', y='usd', data=df_limpo, order=ordem_senioridade, )
plt.title('Boxplot distribuição sálario por experiência')
plt.xlabel('Sálario em USD')
plt.show()
```

```python
#Adicionando comandos palette e hue para adicionar cores nas categorias dos gráficos
ordem_senioridade = ['junior', 'pleno', 'senior', 'executivo']
plt.figure(figsize=(8,5))
sns.boxplot(x='senioridade', y='usd', data=df_limpo, order=ordem_senioridade, palette='Set2', hue='senioridade')
plt.title('Boxplot distribuição sálario por experiência')
plt.xlabel('Sálario em USD')
plt.show()
```

```python
import plotly.express as px
```

```python
#Criando gráfico interativo com plotly
senioridade_media_salario = df_limpo.groupby('senioridade')['usd'].mean().sort_values(ascending=False).reset_index()

fig = px.bar(senioridade_media_salario,
             x='senioridade',
             y='usd',
             title='Média Salarial por Senioridade',
             labels={'senioridade': 'Nível de Senioridade', 'usd': 'Média Salarial Anual (USD)'})

fig.show()
```

```python
remoto_contagem = df_limpo['remoto'].value_counts().reset_index()
remoto_contagem.columns = ['tipo_trabalho', 'quantidade']

fig = px.pie(remoto_contagem,
             names='tipo_trabalho',
             values='quantidade',
             title='Proporção dos tipos de trabalho'

             )

fig.show()
```

```python
remoto_contagem = df_limpo['remoto'].value_counts().reset_index()
remoto_contagem.columns = ['tipo_trabalho', 'quantidade']

fig = px.pie(remoto_contagem,
             names='tipo_trabalho',
             values='quantidade',
             title='Proporção dos tipos de trabalho',
             hole=(0.5)
             )

fig.show()
```

```python
remoto_contagem = df_limpo['remoto'].value_counts().reset_index()
remoto_contagem.columns = ['tipo_trabalho', 'quantidade']

fig = px.pie(remoto_contagem,
             names='tipo_trabalho',
             values='quantidade',
             title='Proporção dos tipos de trabalho',
             hole=(0.5)
             )

fig.update_traces(textinfo='percent+label')
fig.show()
```

```python
df_data_scientist = df_limpo[df_limpo['cargo'] == 'Data Scientist']

salario_medio_pais = df_data_scientist.groupby('residencia')['usd'].mean().sort_values(ascending=False).reset_index()

fig = px.bar(salario_medio_pais,
             x='residencia',
             y='usd',
             title='Média salarial para Data Scientists por país',
             labels={'residencia': 'País', 'usd': 'Média salarial anual (USD)'},
             color='residencia' # Define a cor baseada no país
            )

fig.show()
```
