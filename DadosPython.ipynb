```python
df.rename(columns={
    'ano': 'ano',
    'nivel_de_experiencia': 'senioridade',
    'tipo_de_emprego': 'contrato',
    'cargo': 'cargo',
    'salario': 'salario',
    'moeda_salario': 'moeda',
    'salario_em_usd': 'usd',
    'residencia_do_empregado': 'residencia',
    'Taxa_remoto': 'remoto',
    'localizacao_da_empresa': 'empresa',
    'tamanho_da_empresa': 'tamanho_da_empresa'
}, inplace=True)

df.columns
```

```python
df['senioridade'].value_counts()
```

```python
df['contrato'].value_counts()
```

```python
df['remoto'].value_counts()
```

```python
df['tamanho_da_empresa'].value_counts()
```

```python
senioridade = {
    'SE': 'senior',
    'MI': 'pleno',
    'EN': 'junior',
    'EX': 'executivo'
}
df['senioridade'] = df['senioridade'].replace(senioridade)
df['senioridade'].value_counts()
```

```python
contrato = {
    'FT': 'integral',
    'PT': 'parcial',
    'CT': 'contrato',
    'FL': 'freelancer'
}
df['contrato'] = df['contrato'].replace(contrato)
df['contrato'].value_counts()
```

```python
tamanho_da_empresa = {
    'L': 'grande',
    'S': 'pequena',
    'M':	'media'

}
df['tamanho_da_empresa'] = df['tamanho_da_empresa'].replace(tamanho_da_empresa)
df['tamanho_da_empresa'].value_counts()
```

```python
remoto = {
    0: 'presencial',
    100: 'remoto',
    50: 'hibrido'
}

df['remoto'] = df['remoto'].replace(remoto)
df['remoto'].value_counts()
```
